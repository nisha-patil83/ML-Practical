```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.metrics import r2_score
```

```python
import numpy as np
import pandas as pd
X = np.array([1.1,1.3,1.5,2.0,2.2,2.9,3.0,3.2,3.7,4.0,2.9])
Y = np.array([39000,46000,37700,39800,56600,60100,np.nan,54400,57100,63200,56600])
df = pd.DataFrame({
"Experience":X,
"Salary":Y
})
print(df)
```

```python
plt.plot(X,Y, marker="D")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("original Data - Experience vs Salary")
plt.grid()
plt.show()
```

```python
df.isnull()
```

```python
X =df[["Experience"]]
Y =df["Salary"]
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state=1)
```

```python
X_test.shape
```

```python
Y_train.shape
```

```python
Y_test.shape
```

```python
X_train.shape
```

```python
model = LinearRegression()
```

```python
Y_train_cleaned = Y_train.fillna(Y_train.mean())
model.fit(X_train, Y_train_cleaned)
```

```python
Y_pred = model.predict(X_test)
print("prediction value",Y_pred)
```

```python
r2 = r2_score(Y_test, Y_pred)
print("R2 score",r2)
```

```python
experience =np.array([[5]])
salary = model.predict(experience)
print("Predicted salary for 5 years Experience:", salary[0])
```

```python
plt.plot(X, Y, label="Actual Salary")
plt.plot(X, model.predict(X), label="Regression Line")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Liner Regression - Salary vs Experience")
plt.legend()
plt.grid()
plt.show()
```
